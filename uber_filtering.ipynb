```python
import pandas as pd

df = pd.read_csv('uber_delivery_tips_dataset.csv')
```

```python
result1 = df[(df['earnings'] > 25) & (df['tip'] > 3)]
result1
```

```python
result2 = df[df['part_of_day'].str.contains('Night', case=False)]
result2
```

```python
result3 = df[df['is_weekend'] == True].sort_values(by='tip', ascending=False)
result3
```

```python
result4 = df[(df['distance'].between(3, 8)) & (df['earnings'] > 15)]
result4
```
